```python
import random
list_pets = ["Meg","Duda","Pandora","Joy","Garoa","Mini","Soneca","Honda","Lucky","Bingo"]

#sorteando o nome do PET
sorteioPets = random.choice(list_pets)

# Descobrindo a posição do pet
indice_do_pet = list_pets.index(sorteioPets)

# Usando o pop passando o índice
list_pets.pop(indice_do_pet)

print(f"Pet sorteado e removido: {sorteioPets}")
print(f"Total de pets restantes na lista: {len(list_pets)}")
print(f"Lista atualizada: {list_pets}")
```

```python
# Vamos jogar par ou impar
# Operadores matematicos

# Exemplo, qual é o maior numero
numero1 = int(input('Informe um número: '))
numero2 = int(input('Informe outro número: '))

# laço de condição
# se o numero1 for maior que o numero2 faça
if numero1 > numero2:
  print('O numero 1 é maior que o numero 2')
elif numero1 == numero2:
  print('Os valores são iguais')
else:
  print('O numero 2 é maior do que o numero 1')

```

```python
#Desafio - Criar um jogo de Par ou Impar

numero = int(input('Informe um numero '))

if numero % 2 == 0:
  print('O numero é par')
else:
  print('O numero é impar')
```

```python
#Desafio - criar um jogo par ou impar seguindo as instruções do professor
#1 Pedir para o usuario dois numeros
#2 somar esses dois numeros
#3 descobri se essa soma é par ou impar
# (O DESAFIO ESTA AQUI)
#4 Informar se deu par ou impar (usar if)

n1 = int(input('Informe um numero: '))
n2 = int(input('Informe outro numero'))
soma = n1 + n2

if soma % 2 == 0:
  print('O resultado da soma é par')
else:
  print('O resultado da soma é impar ')
```
